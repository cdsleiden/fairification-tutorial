# Reusable

“Reusability” can be divided into three sub-principles. 

* **R1**. meta(data) are richly described with a plurality of accurate and relevant attributes - for example those mentioned in R1.1, R1.2 and R1.3.
* **R1.1**. (meta)data are released with a clear and accessible data usage license - for example a Creative Commons license(opens in a new tab)
* **R1.2**. (meta)data are associated with detailed provenance – i.e. providing answers to the questions "who, what, when, where, why, and how" using a standard vocabulary such as Dublin Core
* **R1.3**. (Meta)data meet domain-specific community standards – see section 3 under "Interoperability" 

To comply with requirement R1.1, you need to publish your data under a specific licence. A license describes the conditions under which your data or software is (re)usable. The [Creative Commons license picker](https://creativecommons.org/chooser/) may provide useful pointers. If you are not sure about the licence to select, you can always consult a copyright expert. The licence you select ideally permits the reuse of the data.
