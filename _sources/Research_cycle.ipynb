# FAIR principles and the research data life cycle

The [research data life cycle](https://library.ucf.edu/about/departments/scholarly-communication/overview-research-lifecycle) can be divided into different stages: planning research, collecting/reusing data, processing and analysing data, publishing and sharing data, and preserving data. 

* Although datasets can be FAIRified at the moment they are published, it is generally advisable to plan the whole FAIRification process well, and to embed the various activities that are needed within the various staged of the research process. 

* During the planning stage, researchers generally need to write a Data Management plan. In this, it is advisable to stress that the data will be deposited in a repository which assigns a persistent identifier, in fulfilment of the requirements mentioned under "Findable" and "Accessible".

* When collecting existing data, documentation and procedures for data and metadata modelling are of importance. If data is reused, make sure to cite data.
Researchers who develop databases generally start this process by developing a solid data model. During the design of the database, it is advisable to ensure that the names of the properties conform to boradly shared standards whenever possible. When processing and analysing data, documentation is of paramount importance.

* Upon publishing the dataset, researchers generally need to select a license. Defining the protocols for access and choosing a clear usage license is very important at this stage. Choosing an open license evidently enhance the reusability.  

For more information opn the connections between the research data life cycle and FAIR, see "[Final outcomes of the taskforce FAIRification as as Service](https://doi.org/10.5281/zenodo.7546767)".
