# Accessible

Deposit data sets and all the associated metadata in a repository which assigns persistent identifiers is helpful, as it also helps you to meet the requirements stipulated in the principles to enhance the accessibility:

* **A1**. (meta)data are retrievable by their identifier using a standardised communications protocol - for example [HTTPS](https://en.wikipedia.org/wiki/HTTPS)
* **A1.1**. The protocol is open, free, and universally implementable - for example using open standards
* **A1.2**. The protocol allows for an authentication and authorization procedure, where necessary - for example by logging in with a verified account such as your university account
* **A2**. Tetadata are accessible, even when the data are no longer available.

The princple that data ought to be retrievable by their identifier does not necessarily imply that the data should fully be available in open access. It principally means that there needs to be a protocol that users may follow to obtain of the data set. There can be many good reasons for limiting the access to a file. Public accessibility may be difficult because of privacy laws or copyright protection regulations, for example.
Reposities such as [Zenodo](https://zenodo.org/), [4TU.ResearchData](https://data.4tu.nl/) and the [DANS data stations](https://dans.knaw.nl/nl/data-stations/) continue to manage the metadata about a deposited data set, even if the dataset itself becomes inaccessible, for whichever reason. This helps you to comply with requirement A2.
