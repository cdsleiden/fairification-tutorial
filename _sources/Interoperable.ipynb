# Interoperable

As was shown, the various requirements mentioned for the findability and the accessibility can be addressed fairly easily by depositing your data in a repository offering helpful functionalities. Ensuring the interoperability and the reusability of your data will generally demand much more work on your part. The principles mentioned for interoperability concentrate on the practical conditions that may facilitate the reuse of your data. 

How to make your data interoperable also depends on the nature of your data, for example if it is in tabular format or graphical format, or another format such as documents containing interview data. To enhance the interopability a tabular dataset, we advise you to carry out the following activities, in the order that is discussed. The last point, saving your data in an open format, applies to any type of data.

1.	Ensure that your dataset is well-structured 
2.	Describe the conceptual model underlying your data set 
3.	If applicable, replace idiosyncratic variable names with variable names that are widely used in your discipline
4.	Save the data using an open format

These activities will be discussed in more detail below. 

## 1. Ensure that your dataset is well-structured 

At a basic level, it is important to ensure that your data conforms to the central principles of the Tidy Data principles, defined by [Wickham et al.](https://vita.had.co.nz/papers/tidy-data.pdf).  The rules defined as part of the Tidy Data concept will help you to avoid problems associated with messy tabular data. The basic rules can be summarized as follows: 

1.	Each column should contain the values for a single variable in the dataset. 
2.	Each row in the table should contain the observations for a single instance of the entity being described. 
3.	Each cell should contain single or atomic values. 

The third rule is important because, when cells contain multiple values, separated by semi-colons, for example, it can become very difficult to use the data to make calculations. 

The rules for Tidy Data mainly concern the overall structure of your tabular data set. Some additional recommendations can be given about the contents of the data set.  

* An important rule is that you need to be consistent. Use consistent codes for categorical variables, describing locations, organisations or events, for example. When you enter and organise your data in a consistent way from the start of the project, this will save you and your collaborators much time. It means that you won’t have to spend much time harmonising the data at a later stage. 
* Also use a consistent fixed code for any missing values. It is always best to ensure every cell is filled in. Do not use blank cells to represent a missing value, because that can be confusing. Use a code such as “NA” (“Not available”) or NULL in the case of missing values. Avoid numeric values like -999 or 0. 
* Dates are best entered using the ISO 8601 standard. Your are advised, more specifically, to capture dates using the following format: YYYY-MM-DD.
* Do not use colours to indicate properties of data. Represent all data that you need as actual values in the spreadsheet.
* It is advisable to describe the unit of measurement in a data dictionary, rather than in the database itself (see also below). Values should not contain units. All the measurements you place in one column should be in the same unit. 

Well-structured and well-organised data can evidently be reused much more easily. 

## 2. Describe the conceptual model underlying your data set 

As a crucial step in the FAIRifaction process, you are adviced to make a clear and comprehensive list of all the entities and the properties in your dataset. Such an overview of the entities and the properties (or attributes) is often referred to as the conceptual model of the database. The rows in a database table are usually instances of the general topics or entities that are being described. The columns typically describe properties of these entities. These properties specify the aspects or the characteristics of the entities that are captured in the data set (e.g. the title, the material or the year of birth).

You may describe all the entities and properties that are recorded in the database in a text file, in the form a data dictionary, for example. In the case of more complicated databases, you may also make use of a more formal notation technique, such as an Entity Relationship Diagram.

A data dictionary essentially describes the nature and the usage of all the variables described in the various tables in the dataset. Data dictionaries can also be referred to as codebooks.

Such a “data dictionary” might contain:

* The exact variable name as used in the data file
* A version of the variable name that might be used in data visualisations
* A longer explanation of what the variable means
* The measurement units
* Expected minimum and maximum values

Once you have developed a suitable data model, you are also advised to develop a data dictionary which documents the model. This document may contain the following information:

* A list of all the column names used in the data spreadsheet
* A description of the purpose and the contents of these different columns.
* If applicable, give an indication of the units of measurement.
* If applicable, describe the measures that have been taken to ensure the correctness and the consistency of the data
* Explain abbreviations or notational conventions that have been used in the data set.

## 3. Make sure that all the column names are based on terms defined in formal ontologies or vocabularies

Many researchers and institutions have developed shared vocabularies and ontologies to standardise terminology. Such shared vocabularies can help us to make sure that we are talking about the same thing when we exchange knowledge.
The list below mentions a number of widely used vocabularies and ontologies.

* [Wikidata](https://www.wikidata.org/) assigns identifiers to a wide range of entities, including people, locations and organisations
* [VIAF](https://viaf.org/en) (Virtual International Authority File) to identify names, locations, works, and expressions
* Aspects of books can be described using terms from the [Bibliographic Ontology](https://dcmi.github.io/bibo/) and the [FABIO](https://sparontologies.github.io/fabio/current/fabio.html) ontology.
* [FOF](http://xmlns.com/foaf/spec/) (Friend of a Friend) for linking people and information
* [Schema.org](Schema.org) covers many categories such as persons, works and organisations
* [GeoNames](https://www.geonames.org/) defines persistent identifiers to locations, e.g. https://www.geonames.org/2751773/leiden.html.
* [Dublin Core](https://www.dublincore.org/specifications/dublin-core/dcmi-terms/) minimal metadata elements
* [Dublin Core terms](https://www.dublincore.org/specifications/dublin-core/dcmi-terms/) for describing a dataset in a generic way 

The TaDiRAh, [BARTOC](https://bartoc.org/) and [Linked Open Vocabularies](https://lov.linkeddata.es/dataset/lov/) websites also offer valuable overviews of the vocabularies and ontologies that have been developed within specific disciplines. Where possible, try to use terms that have been defined in these existing vocabularies and ontologies in your own data set. 

## 4. Save the data using an open format

The reusability of the data may occasionally be complicated by the fact that the data have been stored using a so-called proprietary format, i.e. a format which can only be used in specific applications and whose specifications are not publicly available. If the data are stored in such proprietary formats, it may be difficult to guarantee the long term accessibility. Open formats, such as JSON, comma separated values files or plain txt files help you to improve the longevity of your data. Non-proprietary file formats never require any special software. 

The activities that have been discussed in this section may sound very labor-intensive. When you have gone through this process, however, your data should meet the requirements stated in first two principles defining the “interoperability”. 

* **I1**. (Meta)data use a formal, accessible, shared, and broadly applicable language for knowledge representation.
* **I2**. (Meta)data use vocabularies that follow FAIR principles

The first requirement implies that data should be saved using an open format. The second requirement essentially asks researchers to work with variable names taken from well-documented vocabularies and ontologies. Incidentally, the activities discussed in this section also help you to meet requirement R1.3, which states that “(Meta)data meet domain-relevant community standards”.
