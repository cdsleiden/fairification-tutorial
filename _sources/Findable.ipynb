## Findable

To ensure that your data can be found, you need to deposit their data sets and all the associated metadata in a repository which assigns persistent identifiers. This activity will help you to meet the four subprinciples defined for findabilty. Even if you are not ready to submit your data to a repository it is good to already think about which one to use, so that you can see what the repository requires and explore how others have prepared their data and metadata. 

* **F1**. (meta)data are assigned a globally unique and persistent identifier - for example a digital object identifier (DOI)
* **F2**. data are described with rich metadata - for example research subject and domain
* **F3**. metadata clearly and explicitly include the identifier of the data it describes - for example a repository metadata record with an explicit link to the data
* **F4**. (meta)data are registered or indexed in a searchable resource - for example a data repository

The following data repositories assign persistent identifiers to submitted datasets:
* A domain-specific [data station at DANS](https://dans.knaw.nl/nl/data-stations/) 
* [4TU.ResearchData](https://data.4tu.nl/)
* [Zenodo](https://zenodo.org/)
* [B2SHARE](https://eudat.eu/service-catalogue/b2share-0)
* [Figshare](https://figshare.com/)

Once a specific data repository has been selected, the data set can be submitted, together with the metadata describing this data set. In the context of data management, metadata is structural information about a data set which describes characteristics such as the quality, the format and the contents. Most repositories require a minimum set of metadata, such as name of the creator, the title and the year of creation. Check what kind of metadata the repository you choose demands. Remember that the effort you put into metadata will contribute to the findability of your dataset.

As mentioned, datasets need to be deposited in repositories that assign persistent identifiers (PIDs) to ensure that online references to publications, research data, and persons remain available in the future. A PID is a specific type of a Uniform Resource Identifier (URI), which is managed by an organisation that links a persistent identification code with the most recent Uniform Resource Locator (URL).
