# Introduction


This tutorial explains the various steps that can be taken if you aim publish your data set in accordance with the FAIR principles. FAIR stands for **Findable**, **Accessible**, **Interoperable** and **Reusable**. The FAIR guiding principles were published as a scientific article in 2016. The article is available in [open access](https://doi.org/10.1038/sdata.2016.18). The formulation of the guidelines were motivated by the conviction that when data are stored, maintain and shared well, these can also be reused and repurposed by others for the creation of new knowledge.
